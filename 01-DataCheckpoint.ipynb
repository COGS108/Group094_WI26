## Research Question

How do weather conditions affect the rate of traffic violations involving autonomous vehicles compared to human-driven vehicles in California?

Specifically, we examine whether autonomous vehicles operating in California have fewer or more traffic violations than human-driven vehicles under different weather conditions such as clear weather, rain, fog, and snow. The independent variable is weather condition (clear, rain, fog, snow), the dependent variable is the rate of traffic violations (including speeding, running red lights, and improper lane changes), and the primary comparison variable is vehicle type (autonomous vs. human-driven).

Using California-specific traffic and weather datasets, this study applies statistical inference methods to compare violation rates across weather conditions and vehicle types. By focusing on California, where autonomous vehicle testing and deployment are heavily regulated and documented, we aim to determine whether autonomous vehicles demonstrate greater safety than human drivers in adverse weather conditions or whether their performance declines under certain environmental factors.



## Background and Prior Work

Autonomous vehicles (AVs) are increasingly deployed on public roads, raising important questions about their safety and compliance with traffic laws relative to human-driven vehicles. While much public discussion has focused on whether AVs reduce crash risk, less attention has been paid to how AVs behave with respect to traffic violations, such as speeding, red-light running, or improper lane changes. Weather conditions present a particularly relevant context for this comparison. Rain, fog, and snow degrade visibility and road friction, increasing driving difficulty for humans and simultaneously stressing the perception and control systems used by autonomous vehicles. Understanding how traffic violation rates vary by weather and vehicle type is therefore critical for evaluating whether autonomous vehicles maintain safer or more lawful behavior under adverse conditions.

**CRASH FOCUSED RESEARCH:**

Most existing research comparing autonomous and human-driven vehicles focuses on crash outcomes rather than traffic violations. A recent large-scale matched case–control study published in Nature Communications compared crash involvement between autonomous and human-driven vehicles while controlling for exposure factors such as location and time of day. This study found that AVs exhibited lower crash rates in certain contexts but emphasized that conclusions depend strongly on how exposure and confounding variables are handled.<sup><a href="#fn1" id="ref1">1</a></sup>

***Limitation:*** However, crash-based analyses face several limitations: crashes are relatively rare events, often underreported or inconsistently categorized, and may not capture everyday rule compliance or near-miss behavior. As noted by the National Highway Traffic Safety Administration (NHTSA), crash datasets are structured around reportable incidents and do not systematically capture non-crash traffic violations or routine enforcement actions.<sup><a href="#fn7" id="ref7">7</a></sup>

Similarly, Kusano et al. analyzed Waymo’s rider-only crash data and compared it to human-driver benchmarks, concluding that AV crash rates were comparable to or lower than human baselines under many conditions, though the authors noted limitations related to operating domains and reporting standards.<sup><a href="#fn2" id="ref2">2</a></sup> These studies establish methodological precedents, such as rate normalization and contextual matching, that are directly relevant for analyzing traffic violation rates, even though violations themselves were not the primary outcome. Because violations occur far more frequently than crashes, analyzing traffic citations may provide a higher-resolution measure of everyday driving behavior and regulatory compliance, particularly in varied environmental conditions.

**WEATHER AND SENSOR RESEARCH:**

A parallel body of work examines how adverse weather conditions affect autonomous vehicle perception and performance, providing insight into mechanisms that may influence traffic violations. Empirical studies of on-road autonomous systems have demonstrated that LiDAR and camera-based detection performance degrades measurably under rain, fog, and snow, leading to reduced object detection range and increased uncertainty.<sup><a href="#fn3" id="ref3">3</a></sup> Comprehensive review articles further document how adverse weather challenges sensor fusion, lane detection, and object tracking, and describe mitigation strategies such as radar integration and weather-robust perception models.<sup><a href="#fn4" id="ref4">4</a></sup> These findings suggest that even if AVs are conservative by design, degraded perception in poor weather may lead to behaviors that increase the likelihood of certain violations (e.g., improper lane positioning or failure to proceed through intersections efficiently).

***Limitation:*** However, these studies are largely conducted in controlled experimental settings or simulation environments. They typically measure detection range, object classification accuracy, or system disengagement frequency rather than legally defined traffic violations. As the California Department of Motor Vehicles (DMV) notes in its autonomous vehicle disengagement reporting program, disengagements reflect system limitations but do not necessarily indicate illegal driving behavior.<sup><a href="#fn8" id="ref8">8</a></sup> Thus, while perception degradation is documented, its real-world impact on enforceable traffic violations remains understudied.

**ENFORCEMENT AND REPORTING:**

Evidence that autonomous vehicles accumulate real-world traffic citations further motivates direct study of violations rather than crashes alone. In 2024, Waymo vehicles operating in San Francisco received hundreds of traffic tickets, primarily related to parking and obstruction, demonstrating that AVs are subject to, and do incur, formal violations under current enforcement regimes.<sup><a href="#fn5" id="ref5">5</a></sup>

***Limitation:*** While many of these citations are not moving violations, they highlight that AV behavior can conflict with traffic regulations in practice. Reporting has also noted ambiguity in how law enforcement handles moving violations committed by driverless vehicles, with California introducing new legislation (effective in 2026) to clarify enforcement procedures for autonomous systems.<sup><a href="#fn6" id="ref6">6</a></sup> This regulatory context suggests that violation data may be incomplete or unevenly recorded across vehicle types, an important consideration for statistical inference.
This regulatory transition creates a methodological challenge - comparisons between AVs and human drivers must account for potential enforcement bias, differences in citation practices, and uneven exposure across geographic regions.

**RESEARCH GAP:**

Finally, existing data infrastructures shape what types of comparative analyses are currently feasible. The National Highway Traffic Safety Administration’s Standing General Order (SGO) requires standardized reporting of crashes involving automated driving systems, providing a structured national dataset for AV safety analysis but excluding non-crash violations.<sup><a href="#fn7" id="ref7">7</a></sup> At the state level, California DMV disengagement reports offer detailed information about AV testing and deployment conditions, which researchers often use to study system limitations and environmental challenges, though these reports do not directly capture violations.<sup><a href="#fn8" id="ref8">8</a></sup> Together, prior work reveals a gap in the literature: while crashes and disengagements have been studied extensively, comparative traffic violation rates across weather conditions and vehicle types remain largely unexplored, motivating the present project.

**This project fills a distinct gap by:**

*   Shifting the outcome variable from crashes to traffic violations (a
higher-frequency behavioral metric).
*   Incorporating weather as a central explanatory variable rather than a background control.
* Using California-specific enforcement data to leverage the state’s unique concentration of AV deployment.

By focusing on violation rates rather than crashes alone, this study provides a complementary perspective on autonomous vehicle safety - one centered on lawful behavior and regulatory compliance in everyday driving contexts, especially under adverse weather conditions.

```python

# =============================================================================
# CALIFORNIA TRAFFIC COLLISION DATA (SWITRS) — DATA WRANGLING
# =============================================================================

# ── 0. IMPORTS & DIRECTORY SETUP ─────────────────────────────────────────────
import os
import sqlite3
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
from scipy import stats

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.float_format", "{:.2f}".format)

for d in ["data/00-raw", "data/01-interim", "data/02-processed"]:
    os.makedirs(d, exist_ok=True)
print("✓ Directories ready")

# =============================================================================
# STEP 1 — LOAD DATASET
# =============================================================================
print("\n" + "="*60)
print("STEP 1 — LOAD DATASET")
print("="*60)

# Locate the SQLite file (handles filenames with/without trailing spaces)
import glob as _glob

_candidates = (
    _glob.glob("data/00-raw/*.sqlite") +
    _glob.glob("data/00-raw/*.sqlite ") +   # trailing-space variant
    _glob.glob("data/00-raw/switrs*")
)
_candidates = [p for p in _candidates if os.path.isfile(p)]

if not _candidates:
    raise FileNotFoundError(
        "switrs.sqlite not found in data/00-raw/.\n"
        "Download from https://www.kaggle.com/datasets/alexgude/"
        "california-traffic-collision-data-from-switrs\n"
        "and place the .sqlite file in data/00-raw/"
    )

db_path = _candidates[0]

print(f"✓ Found SQLite database: {db_path}")

# Connect to the SQLite database
conn = sqlite3.connect(db_path)

# Show available tables
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", conn)
print(f"  Tables available: {tables['name'].tolist()}")

# Read only the first 150k rows from each table
LIMIT = 150_000
print(f"\n⏳ Loading tables (LIMIT {LIMIT:,} rows each) …")

df_case_ids   = pd.read_sql_query(f"SELECT * FROM case_ids   LIMIT {LIMIT}", conn)
df_collisions = pd.read_sql_query(f"SELECT * FROM collisions LIMIT {LIMIT}", conn)
df_parties    = pd.read_sql_query(f"SELECT * FROM parties    LIMIT {LIMIT}", conn)
df_victims    = pd.read_sql_query(f"SELECT * FROM victims    LIMIT {LIMIT}", conn)

# Close the connection
conn.close()

print(f"  case_ids   : {df_case_ids.shape[0]:>7,} rows × {df_case_ids.shape[1]} cols")
print(f"  collisions : {df_collisions.shape[0]:>7,} rows × {df_collisions.shape[1]} cols")
print(f"  parties    : {df_parties.shape[0]:>7,} rows × {df_parties.shape[1]} cols")
print(f"  victims    : {df_victims.shape[0]:>7,} rows × {df_victims.shape[1]} cols")

# Merge the tables into a single DataFrame
print("\n⏳ Merging tables on 'case_id' …")
data = (
    df_collisions
    .merge(df_case_ids, on="case_id", how="left")
    .merge(df_parties,  on="case_id", how="left")
    .merge(df_victims,  on="case_id", how="left")
)

print(f"✓ Merged dataset: {data.shape[0]:,} rows × {data.shape[1]} columns")

# Show the first 5 rows of the merged DataFrame
print("\n=== HEAD OF MERGED DATA ===")
display(data.head())

# Save raw snapshot
data.to_parquet("data/00-raw/switrs_raw.parquet", index=False)
print("\n✓ Raw snapshot saved → data/00-raw/switrs_raw.parquet")

# Keep a working copy
df_raw = data.copy()

# =============================================================================
# STEP 2 — TIDY DATA CHECK
# =============================================================================
print("\n" + "="*60)
print("STEP 2 — TIDY DATA CHECK")
print("="*60)

print("\nColumn dtypes:")
print(df_raw.dtypes.to_string())

# Check for wide-format / stacked repeated-measure columns
repeated_pattern = [c for c in df_raw.columns
                    if any(c.endswith(str(i)) for i in range(1, 8))]
if repeated_pattern:
    print(f"\n⚠  Potential wide-format columns detected: {repeated_pattern[:10]} …")
    print("   → These may need to be pivoted to long/tidy format.")
else:
    print("\n✓ No wide-format (repeated-measure) suffix columns detected.")

# Tidy criteria
print("""
Tidy-data assessment:
  1. Each variable forms a column           ✓  (one column per attribute)
  2. Each observation forms a row           ✓  (each row = one person/party/victim
                                                involved in a collision event)
  3. Each type of observational unit forms
     its own table                          ✓  (collision, party, victim tables exist;
                                                merged here for convenience)

  The merged DataFrame is effectively in long format with one row per
  (collision × party × victim) combination.  This is standard for
  relational collision datasets and is considered tidy for our analysis.
""")

# =============================================================================
# STEP 3 — DATASET SIZE
# =============================================================================
print("="*60)
print("STEP 3 — DATASET SIZE")
print("="*60)

n_rows, n_cols = df_raw.shape
mem_mb = df_raw.memory_usage(deep=True).sum() / 1e6
print(f"  Rows        : {n_rows:>12,}")
print(f"  Columns     : {n_cols:>12,}")
print(f"  Memory usage: {mem_mb:>11.1f} MB")

# =============================================================================
# STEP 4 — MISSING DATA ANALYSIS
# =============================================================================
print("\n" + "="*60)
print("STEP 4 — MISSING DATA ANALYSIS")
print("="*60)

miss = (
    df_raw.isnull()
    .sum()
    .rename("n_missing")
    .to_frame()
    .assign(pct_missing=lambda x: 100 * x["n_missing"] / n_rows)
    .sort_values("pct_missing", ascending=False)
)
miss_cols = miss[miss["n_missing"] > 0]
print(f"\nColumns with ANY missing values: {len(miss_cols)} / {n_cols}")
print(miss_cols.to_string())

# ── 4a. Visualise missingness ─────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

top_miss = miss_cols.head(30)
axes[0].barh(top_miss.index[::-1], top_miss["pct_missing"][::-1], color="#d62728")
axes[0].set_xlabel("% Missing")
axes[0].set_title("Top-30 Columns by % Missing")
axes[0].axvline(50, color="gray", linestyle="--", lw=1, label="50 %")
axes[0].legend()

axes[1].hist(miss["pct_missing"], bins=30, color="#1f77b4", edgecolor="white")
axes[1].set_xlabel("% Missing")
axes[1].set_ylabel("Number of Columns")
axes[1].set_title("Distribution of Missingness Across All Columns")

plt.tight_layout()
plt.savefig("data/01-interim/missingness_overview.png", dpi=120)
plt.show()
print("✓ Saved → data/01-interim/missingness_overview.png")

# ── 4b. Systematic / MAR / MCAR investigation ────────────────────────────────
print("\n── Systematic missingness investigation ──")

# Build a severity flag to probe whether missingness correlates with injury/death
killed_col  = next((c for c in df_raw.columns if "killed_victims" in c), None)
injured_col = next((c for c in df_raw.columns if "injured_victims" in c), None)

if killed_col and injured_col:
    df_raw["_severe_collision"] = (
        (pd.to_numeric(df_raw[killed_col],  errors="coerce").fillna(0) > 0) |
        (pd.to_numeric(df_raw[injured_col], errors="coerce").fillna(0) > 0)
    ).astype(int)

    for col in miss_cols.head(5).index:
        miss_by_sev = df_raw.groupby("_severe_collision")[col].apply(
            lambda s: s.isnull().mean() * 100
        ).rename({0: "non-severe", 1: "severe"})
        print(f"\n  '{col}' — % missing by collision severity:")
        print(miss_by_sev.to_string())

# Chi-square: is road_surface more often missing at night?
if "road_surface" in df_raw.columns and "collision_time" in df_raw.columns:
    df_tmp = df_raw[["road_surface", "collision_time"]].copy()
    t = pd.to_numeric(df_tmp["collision_time"], errors="coerce")
    df_tmp["night"] = t.between(2200, 2359) | t.between(0, 559)
    df_tmp["rs_missing"] = df_tmp["road_surface"].isnull()
    ct = pd.crosstab(df_tmp["night"], df_tmp["rs_missing"])
    chi2, p, *_ = stats.chi2_contingency(ct)
    print(f"\n  Chi-square (road_surface missing ~ night-time): χ²={chi2:.2f}, p={p:.4f}")
    if p < 0.05:
        print("  → Missingness is NON-RANDOM (MAR): night-time collisions differ.")
    else:
        print("  → No significant association; may be MCAR for this pairing.")

print("""
Summary — missingness mechanism interpretation:
  • Columns >80% missing  → structural absence (e.g. victim/pedestrian detail
    only recorded when that party type is present) — MNAR / structurally missing.
  • Medium-missingness columns (~20–80%) — tend to associate with severity or
    circumstance → Missing At Random (MAR).
  • Low-missingness columns (<5%) — appear essentially MCAR.
""")

# =============================================================================
# STEP 5 — OUTLIERS & SUSPICIOUS ENTRIES
# =============================================================================
print("="*60)
print("STEP 5 — OUTLIERS & SUSPICIOUS ENTRIES")
print("="*60)

num_cols_of_interest = [c for c in [
    "killed_victims", "injured_victims", "party_count",
    "pedestrian_killed_count", "pedestrian_injured_count",
    "bicycle_killed_count", "bicycle_injured_count",
    "motorcycle_killed_count", "motorcycle_injured_count",
    "number_killed", "number_injured",
] if c in df_raw.columns]

if num_cols_of_interest:
    desc = df_raw[num_cols_of_interest].describe(percentiles=[.01,.25,.5,.75,.99]).T
    print("\nDescriptive stats (key numeric columns):")
    print(desc.to_string())

# Flag extreme fatality rows
kv_col = next((c for c in df_raw.columns if c in ("killed_victims","number_killed")), None)
if kv_col:
    extreme = df_raw[pd.to_numeric(df_raw[kv_col], errors="coerce") > 10]
    print(f"\n  Rows with {kv_col} > 10 : {len(extreme):,}")
    if len(extreme):
        show_cols = [c for c in ["case_id","collision_date","killed_victims",
                                  "number_killed","injured_victims"] if c in extreme.columns]
        print(extreme[show_cols].head(10))

# IQR-based outlier count
print("\n  IQR-based outlier counts (> Q3+3×IQR or < Q1-3×IQR):")
for col in num_cols_of_interest:
    s = pd.to_numeric(df_raw[col], errors="coerce").dropna()
    Q1, Q3 = s.quantile(0.25), s.quantile(0.75)
    IQR = Q3 - Q1
    n_out = ((s < Q1 - 3*IQR) | (s > Q3 + 3*IQR)).sum()
    if n_out:
        print(f"    {col:<45s}: {n_out:>6,} outliers  (max={s.max():.0f})")

# Date sanity check
date_col = next((c for c in df_raw.columns if "collision_date" in c), None)
if date_col:
    df_raw["_cdate"] = pd.to_datetime(df_raw[date_col], errors="coerce")
    future   = df_raw[df_raw["_cdate"] > pd.Timestamp.today()]
    pre_1960 = df_raw[df_raw["_cdate"] < pd.Timestamp("1960-01-01")]
    print(f"\n  Dates in the future     : {len(future):,}")
    print(f"  Dates before 1960       : {len(pre_1960):,}")

# Boxplots for key count variables
if num_cols_of_interest:
    ncols_plot = min(4, len(num_cols_of_interest))
    fig, axes = plt.subplots(1, ncols_plot, figsize=(4*ncols_plot, 4))
    if ncols_plot == 1:
        axes = [axes]
    for ax, col in zip(axes, num_cols_of_interest[:ncols_plot]):
        clean_vals = pd.to_numeric(df_raw[col], errors="coerce").dropna()
        ax.boxplot(clean_vals, vert=True, patch_artist=True,
                   boxprops=dict(facecolor="#aec7e8"))
        ax.set_title(col, fontsize=9)
        ax.set_ylabel("Count")
    plt.suptitle("Outlier Check — Key Count Variables", y=1.02)
    plt.tight_layout()
    plt.savefig("data/01-interim/outlier_boxplots.png", dpi=120)
    plt.show()
    print("✓ Saved → data/01-interim/outlier_boxplots.png")

# =============================================================================
# STEP 6 — DATA CLEANING
# =============================================================================
print("\n" + "="*60)
print("STEP 6 — DATA CLEANING")
print("="*60)

df = df_raw.copy()

# ── 6a. Drop columns with ≥90 % missing — structural absence, not recoverable ─
THRESHOLD_DROP_COL = 0.90
cols_to_drop = miss[miss["pct_missing"] >= THRESHOLD_DROP_COL * 100].index.tolist()
cols_to_drop = [c for c in cols_to_drop if c in df.columns]
print(f"\n  Dropping {len(cols_to_drop)} columns with ≥{int(THRESHOLD_DROP_COL*100)}% missing:")
print("  ", cols_to_drop)
df.drop(columns=cols_to_drop, inplace=True)

# ── 6b. Remove internal helper columns ────────────────────────────────────────
helper_cols = [c for c in df.columns if c.startswith("_")]
df.drop(columns=helper_cols, errors="ignore", inplace=True)

# ── 6c. Fix column types ──────────────────────────────────────────────────────
if date_col and date_col in df.columns:
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

time_col = next((c for c in df.columns if "collision_time" in c), None)
if time_col:
    df[time_col] = pd.to_numeric(df[time_col], errors="coerce")

# Downcast numerics to save memory
for col in df.select_dtypes("float64").columns:
    df[col] = df[col].astype("float32")
for col in df.select_dtypes("int64").columns:
    df[col] = df[col].astype("int32")

# ── 6d. Fill count columns with 0 where NaN (MNAR — no entry means 0 events) ─
#  Justification: victim/party count fields are left blank when the count is 0,
#  so NaN is semantically equivalent to 0.  Replacing with 0 preserves the
#  full record rather than creating spurious row deletions.
#  We explicitly target only numeric count columns (suffix _count, _killed, _injured)
#  and party_count to avoid touching categorical columns like pedestrian_action.
count_fill_cols = [c for c in df.columns
                   if c.endswith(("_count", "_killed", "_injured"))
                   or c == "party_count"]
# Keep only those that are actually numeric (or fully null/numeric)
count_fill_cols_present = [
    c for c in count_fill_cols
    if c in df.columns and pd.api.types.is_numeric_dtype(df[c])
]
df[count_fill_cols_present] = df[count_fill_cols_present].fillna(0)
print(f"\n  Count columns filled with 0 (MNAR→0): {count_fill_cols_present}")

# ── 6e. Drop rows that are entirely null (true empty records) ─────────────────
before = len(df)
df.dropna(how="all", inplace=True)
after  = len(df)
print(f"\n  dropna(how='all') removed : {before - after:,} fully-null rows")
print(f"  Remaining rows            : {after:,}")

# ── 6f. Remove impossible / bad dates ─────────────────────────────────────────
if date_col and date_col in df.columns:
    before = len(df)
    df = df[(df[date_col] >= "1960-01-01") &
            (df[date_col] <= pd.Timestamp.today())]
    print(f"\n  Date-range filter removed : {before - len(df):,} rows")

# ── 6g. Final shape & remaining NaNs ──────────────────────────────────────────
print(f"\nFinal cleaned dataset : {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Remaining NaN cells   : {df.isnull().sum().sum():,}")

# Save interim
df.to_parquet("data/01-interim/switrs_interim.parquet", index=False)
print("✓ Interim file saved → data/01-interim/switrs_interim.parquet")

# =============================================================================
# STEP 7 — WRITE PROCESSED FILE
# =============================================================================
print("\n" + "="*60)
print("STEP 7 — WRITING PROCESSED FILE")
print("="*60)

df.to_parquet("data/02-processed/switrs_clean.parquet", index=False)
print(f"✓ Processed file saved → data/02-processed/switrs_clean.parquet")
print(f"  Rows    : {len(df):,}")
print(f"  Columns : {df.shape[1]}")
print(f"  Size    : {df.memory_usage(deep=True).sum()/1e6:.1f} MB")

# =============================================================================
# STEP 8 — SUMMARY STATISTICS (KEY VARIABLES)
# =============================================================================
print("\n" + "="*60)
print("STEP 8 — SUMMARY STATISTICS (KEY VARIABLES)")
print("="*60)

# Collision severity distribution
sev_col = next((c for c in df.columns if "collision_severity" in c.lower() or
                "severity" in c.lower()), None)
if sev_col:
    print(f"\nCollision Severity  ({sev_col}):")
    print(df[sev_col].value_counts(dropna=False).to_string())

# Year-over-year collision counts
if date_col and date_col in df.columns:
    df["year"] = df[date_col].dt.year
    yearly = df.groupby("year").size().rename("collisions")
    print("\nCollisions per Year (merged dataset):")
    print(yearly.to_string())

    fig, ax = plt.subplots(figsize=(12, 4))
    yearly.plot(kind="bar", ax=ax, color="#1f77b4", edgecolor="white")
    ax.set_title("Annual Collision Count — California (SWITRS, first 150k rows)")
    ax.set_xlabel("Year")
    ax.set_ylabel("Number of Rows")
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x):,}"))
    plt.tight_layout()
    plt.savefig("data/01-interim/yearly_collisions.png", dpi=120)
    plt.show()
    print("✓ Saved → data/01-interim/yearly_collisions.png")

# Fatality & injury totals
kv = next((c for c in df.columns if c in ("killed_victims","number_killed")), None)
iv = next((c for c in df.columns if c in ("injured_victims","number_injured")), None)
if kv:
    print(f"\nTotal killed  ({kv})  : {int(pd.to_numeric(df[kv], errors='coerce').sum()):,}")
if iv:
    print(f"Total injured ({iv}) : {int(pd.to_numeric(df[iv], errors='coerce').sum()):,}")

# Top primary collision factors
pcf_col = next((c for c in df.columns
                if "primary_collision_factor" in c or "pcf_violation" in c.lower()), None)
if pcf_col:
    print(f"\nTop 10 Primary Collision Factors  ({pcf_col}):")
    print(df[pcf_col].value_counts().head(10).to_string())

# Hour-of-day distribution
if time_col and time_col in df.columns:
    df["hour"] = (pd.to_numeric(df[time_col], errors="coerce")
                  .fillna(0) // 100).astype(int).clip(0, 23)
    hourly = df.groupby("hour").size()
    fig, ax = plt.subplots(figsize=(12, 4))
    hourly.plot(kind="bar", ax=ax, color="#ff7f0e", edgecolor="white")
    ax.set_title("Collisions by Hour of Day (SWITRS dataset)")
    ax.set_xlabel("Hour (24-hr)")
    ax.set_ylabel("Count")
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{int(x):,}"))
    plt.tight_layout()
    plt.savefig("data/01-interim/hourly_collisions.png", dpi=120)
    plt.show()
    print("✓ Saved → data/01-interim/hourly_collisions.png")

# Party type breakdown
party_type_col = next((c for c in df.columns if "party_type" in c.lower()), None)
if party_type_col:
    print(f"\nParty Types  ({party_type_col}):")
    print(df[party_type_col].value_counts(dropna=False).head(10).to_string())

# Victim role breakdown
victim_role_col = next((c for c in df.columns if "victim_role" in c.lower()), None)
if victim_role_col:
    print(f"\nVictim Roles  ({victim_role_col}):")
    print(df[victim_role_col].value_counts(dropna=False).head(10).to_string())

print("\n" + "="*60)
print("WRANGLING COMPLETE")
print("="*60)
print(f"""
  data/00-raw/switrs_raw.parquet          ← full merged raw snapshot
  data/01-interim/switrs_interim.parquet  ← after type-fixing & count fill
  data/02-processed/switrs_clean.parquet  ← final clean dataset
""")
```

#### Important Metrics

**Weather** is recorded as a categorical variable with seven possible values: Clear, Cloudy, Raining, Snowing, Fog, Other, and Wind. This is the main independent variable in the analysis. The weather condition reflects what was reported at the time of the collision by the AV manufacturer on the DMV form. The distribution is highly imbalanced. About 84% of collisions occurred in clear weather, roughly 11% in cloudy conditions, and about 5% in rain. Fog and "Other" each account for fewer than 1% of cases. There are no recorded collisions in snow or wind conditions. This imbalance is important because most AV testing in California takes place in San Francisco where snow essentially doesn't occur and it is relatively rare for dense fog to occur during testing periods. Thus, the dataset contains very few observations in poor weather categories. Because there are only 5 fog cases and no snow cases at all, we really can't say much about how AVs perform in those conditions. Any conclusions drawn from such tiny (or nonexistent) samples would be weak and should be treated very cautiously.

**Autonomous Mode vs. Conventional Mode** is recorded using two binary columns (Yes or blank) to indicate whether the vehicle's self-driving system was in control at the time of the collision or whether a human safety driver was operating the vehicle. This is the central variable for comparing system-driven behavior to human-driven behavior within the same AV fleet. In this dataset, 358 collisions (55%) occurred while the vehicle was in autonomous mode and 284 (44%) occurred in conventional mode, with 4 records missing this information. The distribution is fairly balanced, which is helpful because it gives us a meaningful number of observations in both conditions. That said, it is important to interpret "conventional mode" correctly. In this context, it does not refer to an average human driver. It refers to a trained safety driver employed by an AV company who is operating a vehicle equipped with autonomous technology. These drivers are typically more experienced, more attentive, and specifically trained for this role. As a result, comparisons between autonomous and conventional mode in this dataset likely underestimate the difference between AV performance and the broader driving population.

**Vehicle Damage** is recorded as a categorical variable with five levels: None, Minor, Moderate, Major, and Unknown. The damage level is reported by the AV company itself, following the DMV damage scale, rather than by an independent appraiser. "Minor" typically refers to cosmetic damage such as scratches or small dents with no structural impact. "Moderate" indicates that some structural components were affected, but the vehicle is still drivable. "Major" reflects significant structural damage and may mean the vehicle is no longer drivable. In this dataset, 73% of collisions resulted in minor damage, 15% in moderate damage, 9% in no damage at all, and only 2.3% in major damage. This distribution suggests that most AV-related collisions are relatively low-speed and low-severity incidents. For the purposes of our research question, damage severity acts as a practical measure of collision intensity. It allows us to examine whether certain conditions, such as adverse weather, are associated with more severe outcomes.

**Type of Collision** is recorded as a categorical variable with eight possible categories: Head-On, Sideswipe, Rear End, Broadside, Hit Object, Overturned, Vehicle/Pedestrian, and Other. This variable captures the physical layout or geometry of the crash. Rear-end collisions are the most common type. They account for 18% of cases from the AV's perspective and 32% from the other vehicle's perspective, which suggests that AVs are often being rear-ended rather than causing the rear-end impact themselves. This matters for the weather analysis because rear-end crashes are closely tied to following distance and braking behavior. In conditions like rain or on slippery roads, stopping distances increase and reaction time becomes more critical.

**Struck by Others** is a binary variable coded as 1 for yes and 0 for no. It indicates whether the AV was hit by another vehicle, rather than the AV striking something else. In this dataset, 84% of collisions involve the AV being the vehicle that was struck. For our research question, this variable helps separate collisions that may be linked to the AV's own driving behavior, which could potentially worsen in adverse weather, from collisions that are primarily the result of human driver behavior under the same conditions.

**Road Surface** is recorded as a categorical variable with four levels: Dry, Wet, Snowy/Icy, and Slippery. It describes the physical condition of the roadway at the time of the collision. Although it is related to the weather variable, it captures something slightly different. For example, the road may still be wet after rainfall has stopped, or it may remain dry during cloudy conditions. This variable complements the weather data by giving us a more precise way to assess how environmental factors influence collision patterns.

**Date of Accident and Time of Accident** are recorded separately as a date string (for example, "6/25/2021") and a time string (for example, "2:52"), along with separate AM/PM indicator columns. The dataset covers incidents from January 2019 through December 2024. Time is relevant because certain weather conditions like fog are more likely at specific times of day, and lighting conditions interact with weather to affect driving difficulty.

**Latitude and Longitude** are recorded in decimal degrees using the WGS84 coordinate system. Latitude ranges from about 32.66 (San Diego) up to 37.84 (northern San Francisco), while longitude ranges from roughly −122.51 to −116.27. The data are heavily concentrated around San Francisco (lat ≈ 37.77, lon ≈ −122.42), reflecting where most AV testing occurs. Location matters because weather patterns vary across regions, and we may need to account for geographic concentration to avoid attributing location-based effects to weather alone.

#### Concerns and Limitations

**Self-reported by AV companies.** Every record comes directly from the autonomous vehicle companies themselves. The company writes the accident summary, decides how severe the damage was, and reports whether the car was in autonomous or conventional mode. There is no outside investigator double-checking these details. AV manufacturers have obvious financial and reputational reasons to frame incidents favorably. So while 84% of collisions are reported as the AV being "struck by others," we should keep in mind that this conclusion is based on the company's own account.

**Extreme geographic concentration.** About 75% of the 646 records come from San Francisco, and nearly all remaining cases are from nearby Bay Area cities. San Francisco has a mild Mediterranean climate — it essentially never snows, dense street-level fog is uncommon, and rainfall is moderate. As a result, the dataset contains almost no observations for the weather conditions most relevant to our research question: snow has 0 records, fog has 5, and rain only 31. Any conclusions about AV performance in adverse weather will be based on very limited evidence, and results will not generalize to regions with harsher climates.

**No human-driver comparison group in this dataset.** This dataset only includes collisions involving autonomous vehicles. To compare AV collision rates to human-driver collision rates across weather conditions, we would need to combine this with a separate human-driver dataset and normalize by exposure (e.g., collisions per million miles driven per weather condition). Without that normalization, raw counts are misleading — more collisions in clear weather doesn't mean clear weather is more dangerous; it means more driving happens when conditions are clear.

**No traffic violation data.** Our research question focuses on violation rates, but this dataset only records collisions, not violations. A collision is just one possible outcome of unsafe driving, and many violations never lead to a crash. We would need to treat collisions as an imperfect proxy for unsafe behavior or find an additional violation-specific dataset.

**"Conventional mode" does not equal a normal human driver.** In this dataset, conventional mode refers to trained safety drivers employed by AV companies, not typical members of the public. These drivers are selected, trained, and monitored for this role. Any differences between autonomous and conventional mode should be interpreted as a controlled comparison between software and trained professionals, not software versus average drivers.

## Important Metrics

**Weather Condition**  
Weather is recorded as a categorical variable describing environmental conditions at the time of the collision (e.g., clear, rain, fog, snow). This variable is central to the research question because it allows comparison of collision patterns under different driving conditions. However, severe weather categories such as snow occur infrequently in California, which may limit statistical power for those conditions.

**Road Surface Condition**  
Road surface is typically categorized as dry, wet, snowy/icy, or slippery. This variable complements weather data by capturing residual effects of precipitation (e.g., wet roads after rain has stopped). Road surface conditions are directly related to braking distance and vehicle control, making them important indicators of driving difficulty.

**Collision Severity**  
Severity is measured through indicators such as number of fatalities, number of injuries, or severity categories (e.g., property damage only vs. injury collisions). This metric provides insight into the seriousness of crashes and may reflect how dangerous conditions are under different weather scenarios.

**Number of Victims (Injured/Killed)**  
Counts of injured and killed individuals quantify the human impact of collisions. These variables allow analysis beyond simple crash frequency and can indicate whether adverse weather leads to more severe outcomes.

**Primary Collision Factor**  
This variable records the main cause attributed by investigating officers (e.g., speeding, unsafe turning, failure to yield). It serves as a proxy for unsafe driving behavior and helps distinguish collisions caused by driver actions from those primarily due to environmental factors.

**Date and Time of Collision**  
Temporal variables enable analysis of time-of-day effects, seasonal trends, and correlations between weather conditions and traffic density.

**Location (Latitude, Longitude, County)**  
Geographic variables allow spatial analysis and control for regional differences. Weather patterns vary significantly across California, from coastal fog to inland heat and mountain snow.

**Number of Vehicles Involved**  
This variable indicates collision complexity. Multi-vehicle crashes may be more common under poor visibility or slippery conditions.

## Concerns and Limitations

*Represents collisions only, not all violations**  
The SWITRS dataset records only incidents that resulted in reported crashes. Traffic violations that did not lead to collisions are not included, meaning the dataset may underestimate risky driving behavior.

**Underreporting of minor incidents**  
Property-damage-only collisions or minor incidents may not always be reported to police, leading to bias toward more severe crashes.

**Human-driver dataset is not perfectly comparable to AV dataset**  
SWITRS reflects typical drivers, while the AV dataset includes autonomous operation and trained safety drivers. Differences in driver populations and operating conditions may confound comparisons.

**Geographic and environmental variation**  
California includes diverse climates, but severe weather occurs mainly in specific regions. Most collisions occur in urban areas with mild weather.

**Exposure not measured**  
The dataset does not include miles driven or traffic volume. Higher collision counts in clear weather may reflect more driving rather than greater risk.

**Potential reporting inconsistencies**  
Data are collected across multiple jurisdictions, which may use slightly different reporting standards.

**Missing or incomplete data**  
Some variables may contain missing values or ambiguous entries.

**No direct indicator of autonomous vehicles**  
The dataset does not explicitly identify whether a vehicle was autonomous, so it primarily represents human-driven traffic conditions.
