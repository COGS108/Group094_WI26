```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
df = pd.read_csv('data/00-raw/CA_AV_Collision.csv')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)


# The code of this function is completely generated by Google Gemini from the prompt 
# "can I get code to transform the Weather A 1, Weather A 2, up to Weather G 2 columns to two columns of Weather 1, Weather 2 where their values are now a string of ["Clear", "Cloudy", "Raining", "Snowing", "Fog", "Other", "Wind"]?"
def transform_weather_columns(df):
    """
    Consolidates Weather A-G columns into two columns: Weather 1 and Weather 2.
    """
    weather_mapping = {
    'A': 'Clear',
    'B': 'Cloudy',
    'C': 'Raining',
    'D': 'Snowing',
    'E': 'Fog',
    'F': 'Other',
    'G': 'Wind'
    }
    for suffix in ['1', '2']:
        new_column_name = f'Weather {suffix}'
        # Initialize the new column with None/NaN
        df[new_column_name] = np.nan
        df[new_column_name] = df[new_column_name].astype('object')
        
        for char, label in weather_mapping.items():
            old_col = f'Weather {char} {suffix}'
            
            if old_col in df.columns:
                # Identify rows where this weather condition is marked.
                # We check for common indicators like 1, '1', 'X', or True.
                mask = df[old_col].astype(str).str.lower().isin(['1', '1.0', 'true', 'x', 'yes'])
                
                # Assign the label to the new consolidated column
                df.loc[mask, new_column_name] = label
                
    return df

# The code of this function is completely generated by Google Gemini from the prompt 
# "can I get code to transform the Lighting A 1, Lighting A 2, up to Lighting E 2 columns to two columns of Lighting 1, Lighting 2 where their values are now a string of [Daylight, Dark-Dawn, Dark-Street lights, Dark-No street lights, Dark-Street lights not functioning]?"
def transform_lighting_columns(df):
    """
    Consolidates Lighting A-E columns into two columns: Lighting 1 and Lighting 2.
    """
    lighting_mapping = {
    'A': 'Daylight',
    'B': 'Dark-Dawn',
    'C': 'Dark-Street lights',
    'D': 'Dark-No street lights',
    'E': 'Dark-Street lights not functioning'
    }

    for suffix in ['1', '2']:
        new_column_name = f'Lighting {suffix}'
        
        # Initialize with NaN, but immediately cast to 'object' to allow strings
        df[new_column_name] = np.nan
        df[new_column_name] = df[new_column_name].astype('object')
        
        for char, label in lighting_mapping.items():
            old_col = f'Lighting {char} {suffix}'
            
            if old_col in df.columns:
                # Check for indicators of selection (1, '1', 'X', 'Yes', True)
                mask = df[old_col].astype(str).str.lower().isin(['1', '1.0', 'true', 'x', 'yes'])
                df.loc[mask, new_column_name] = label
                
    return df

# The code of this function is completely generated by Google Gemini from the prompt 
# "consolidate the damage columns to be a single column"
def transform_damage_columns(df):
    """
    Consolidates 'Damage: X' columns into a single 'Damage' column.
    """
    # The new single column to hold the result
    new_column_name = 'Damage'
    
    # Initialize with NaN, but immediately cast to 'object' to allow strings
    df[new_column_name] = np.nan
    df[new_column_name] = df[new_column_name].astype('object')
    
    # Mapping of column name -> target label
    # Based on the headers seen in your image
    damage_mapping = {
        'Damage: Unknown': 'Unknown',
        'Damage: None': 'None',
        'Damage: Minor': 'Minor',
        'Damage: Moderate': 'Moderate',
        'Damage: Major': 'Major'
    }

    for old_col, label in damage_mapping.items():
        if old_col in df.columns:
            # Check for indicators of selection (1, '1', 'X', 'Yes', True)
            # The image shows "Yes" specifically, but we keep the list broad just in case
            mask = df[old_col].astype(str).str.lower().isin(['1', '1.0', 'true', 'x', 'yes'])
            
            # Assign the label (e.g., 'Moderate') to the new consolidated column
            df.loc[mask, new_column_name] = label
            
    return df
# The code of this function is completely generated by Google Gemini from the prompt 
# "can I get code to transform the Type A 1, Type A 2, up to Type H 2 columns to two columns of Type 1, Type 2 where their values are now a string of [Head-on, Side swipe, Rear end, Broadside, Hit object, Overturned, Vehicle/pedestrian, Other]?"
def transform_collision_type_columns(df):
    """
    Consolidates Type A-H columns into two columns: 
    'Collision Type 1' and 'Collision Type 2'.
    """
    
    # Mapping based on the order you provided:
    # A=Head-on, B=Side swipe, C=Rear end, etc.
    collision_mapping = {
        'A': 'Head-on',
        'B': 'Side swipe',
        'C': 'Rear end',
        'D': 'Broadside',
        'E': 'Hit object',
        'F': 'Overturned',
        'G': 'Vehicle/pedestrian',
        'H': 'Other'
    }

    for suffix in ['1', '2']:
        new_column_name = f'Collision Type {suffix}'
        
        # Initialize with NaN, but immediately cast to 'object' to allow strings
        df[new_column_name] = np.nan
        df[new_column_name] = df[new_column_name].astype('object')
        
        for char, label in collision_mapping.items():
            # Matches headers like "Type A 1", "Type B 2" seen in your image
            old_col = f'Type {char} {suffix}'
            
            if old_col in df.columns:
                # Check for indicators of selection (1, '1', 'X', 'Yes', True)
                mask = df[old_col].astype(str).str.lower().isin(['1', '1.0', 'true', 'x', 'yes'])
                
                # Assign the label to the new consolidated column
                df.loc[mask, new_column_name] = label
                
    return df

# The code of this function is completely generated by Google Gemini from the prompt 
# "can I get code to transform the Roadway A 1, Roadway A 2, up to Roadway H 2 columns to two columns of Roadway 1, Roadway 2 where their values are now a string of [Dry, Wet, Snowy-Icy, Slippery]?"
def transform_roadway_columns(df):
    """
    Consolidates Roadway A-D columns into two columns: 
    'Roadway 1' and 'Roadway 2'.
    """
    
    # Mapping based on your provided list:
    # A=Dry, B=Wet, C=Snowy-Icy, D=Slippery
    roadway_mapping = {
        'A': 'Dry',
        'B': 'Wet',
        'C': 'Snowy-Icy',
        'D': 'Slippery'
    }

    for suffix in ['1', '2']:
        new_column_name = f'Roadway {suffix}'
        
        # Initialize with NaN, but immediately cast to 'object' to allow strings
        df[new_column_name] = np.nan
        df[new_column_name] = df[new_column_name].astype('object')
        
        for char, label in roadway_mapping.items():
            # Matches headers like "Roadway A 1", "Roadway B 2"
            old_col = f'Roadway {char} {suffix}'
            
            if old_col in df.columns:
                # Check for indicators of selection (1, '1', 'X', 'Yes', True)
                mask = df[old_col].astype(str).str.lower().isin(['1', '1.0', 'true', 'x', 'yes'])
                
                # Assign the label to the new consolidated column
                df.loc[mask, new_column_name] = label
                
    return df

transform_weather_columns(df)
transform_lighting_columns(df)
transform_damage_columns(df)
transform_collision_type_columns(df)
transform_roadway_columns(df)


cleaned_df = df[['Date Of Accident','Time Of Accident','AM','PM','Zip Code','Autonomous Mode','Conventional Mode',
                 'Latitude','Longitude','Struck by Others','Weather 1','Lighting 1','Damage','Collision Type 1','Collision Type 2',
                 'Roadway 1']]


cleaned_df = cleaned_df.rename(columns={'Weather 1': 'Weather', 'Lighting 1': 'Lighting', 'Roadway 1': 'Road Condition'})

cleaned_df.head(10)
```
