## Abstract

As autonomous vehicles (AVs) move from controlled tests to everyday California roads, understanding how weather affects their collision patterns — and how those patterns compare to human drivers — is a pressing safety question. Rain, fog, and snow degrade the cameras and LiDAR that AVs rely on, yet most prior work focuses on crash rates rather than the types and severity of collisions that actually occur under different weather conditions.

This study uses two California-specific datasets to investigate that gap. The first is the Statewide Integrated Traffic Records System (SWITRS), which provides 339,855 police-reported human-driver collision records spanning 2001–2020 and serves as the human-driver baseline. The second is the California DMV Autonomous Vehicle Collision Reports dataset (646 records, 2019–2024), which documents AV-involved incidents. We conduct exploratory data analysis — including weather-by-violation heatmaps, row-normalized severity comparisons, temporal trend analysis, and hourly/geographic breakdowns — to characterize how collision patterns shift under adverse weather for each vehicle type.

Several findings stand out. For human drivers, clear weather accounts for roughly 74% of collisions, yet fog produces the highest per-collision fatality rate despite its small sample, and the share of speeding-related violations rises sharply under fog (47%) and snow (73%). Collision counts dropped steeply in 2020 due to COVID-19, and collisions peak between 3–6 PM with heavy geographic concentration in urban counties. For AVs, collisions are overwhelmingly low-severity (73% minor damage), approximately 84% involve the AV being struck by another party, and the dataset contains almost no fog or snow cases — limiting direct comparison under the most challenging conditions.

These results partially support the hypothesis that AVs exhibit less severe collision outcomes under normal and moderately adverse conditions, but the comparison arm under fog and snow remains untested due to data scarcity. A critical caveat is that no exposure data (e.g., miles driven) is available for either dataset, so observed counts reflect both true risk and driving activity. Future work should integrate exposure-normalized rates and geographically diverse AV deployment data to draw firmer conclusions.

## Background and Prior Work

Autonomous vehicles (AVs) are increasingly deployed on public roads, raising important questions about their safety and compliance with traffic laws relative to human-driven vehicles. While much public discussion has focused on whether AVs reduce crash risk, less attention has been paid to how AVs behave with respect to traffic violations, such as speeding, red-light running, or improper lane changes. Weather conditions present a particularly relevant context for this comparison. Rain, fog, and snow degrade visibility and road friction, increasing driving difficulty for humans and simultaneously stressing the perception and control systems used by autonomous vehicles. Understanding how traffic violation rates vary by weather and vehicle type is therefore critical for evaluating whether autonomous vehicles maintain safer or more lawful behavior under adverse conditions.

**CRASH FOCUSED RESEARCH:**

Most existing research comparing autonomous and human-driven vehicles focuses on crash outcomes rather than traffic violations. A recent large-scale matched case–control study published in Nature Communications compared crash involvement between autonomous and human-driven vehicles while controlling for exposure factors such as location and time of day. This study found that AVs exhibited lower crash rates in certain contexts but emphasized that conclusions depend strongly on how exposure and confounding variables are handled.<sup><a href="#fn1" id="ref1">1</a></sup>

***Limitation:*** However, crash-based analyses face several limitations: crashes are relatively rare events, often underreported or inconsistently categorized, and may not capture everyday rule compliance or near-miss behavior. As noted by the National Highway Traffic Safety Administration (NHTSA), crash datasets are structured around reportable incidents and do not systematically capture non-crash traffic violations or routine enforcement actions.<sup><a href="#fn7" id="ref7">7</a></sup>

Similarly, Kusano et al. analyzed Waymo’s rider-only crash data and compared it to human-driver benchmarks, concluding that AV crash rates were comparable to or lower than human baselines under many conditions, though the authors noted limitations related to operating domains and reporting standards.<sup><a href="#fn2" id="ref2">2</a></sup> These studies establish methodological precedents, such as rate normalization and contextual matching, that are directly relevant for analyzing traffic violation rates, even though violations themselves were not the primary outcome. Because violations occur far more frequently than crashes, analyzing traffic citations may provide a higher-resolution measure of everyday driving behavior and regulatory compliance, particularly in varied environmental conditions.

**WEATHER AND SENSOR RESEARCH:**

A parallel body of work examines how adverse weather conditions affect autonomous vehicle perception and performance, providing insight into mechanisms that may influence traffic violations. Empirical studies of on-road autonomous systems have demonstrated that LiDAR and camera-based detection performance degrades measurably under rain, fog, and snow, leading to reduced object detection range and increased uncertainty.<sup><a href="#fn3" id="ref3">3</a></sup> Comprehensive review articles further document how adverse weather challenges sensor fusion, lane detection, and object tracking, and describe mitigation strategies such as radar integration and weather-robust perception models.<sup><a href="#fn4" id="ref4">4</a></sup> These findings suggest that even if AVs are conservative by design, degraded perception in poor weather may lead to behaviors that increase the likelihood of certain violations (e.g., improper lane positioning or failure to proceed through intersections efficiently).

***Limitation:*** However, these studies are largely conducted in controlled experimental settings or simulation environments. They typically measure detection range, object classification accuracy, or system disengagement frequency rather than legally defined traffic violations. As the California Department of Motor Vehicles (DMV) notes in its autonomous vehicle disengagement reporting program, disengagements reflect system limitations but do not necessarily indicate illegal driving behavior.<sup><a href="#fn8" id="ref8">8</a></sup> Thus, while perception degradation is documented, its real-world impact on enforceable traffic violations remains understudied.

**ENFORCEMENT AND REPORTING:**

Evidence that autonomous vehicles accumulate real-world traffic citations further motivates direct study of violations rather than crashes alone. In 2024, Waymo vehicles operating in San Francisco received hundreds of traffic tickets, primarily related to parking and obstruction, demonstrating that AVs are subject to, and do incur, formal violations under current enforcement regimes.<sup><a href="#fn5" id="ref5">5</a></sup>

***Limitation:*** While many of these citations are not moving violations, they highlight that AV behavior can conflict with traffic regulations in practice. Reporting has also noted ambiguity in how law enforcement handles moving violations committed by driverless vehicles, with California introducing new legislation (effective in 2026) to clarify enforcement procedures for autonomous systems.<sup><a href="#fn6" id="ref6">6</a></sup> This regulatory context suggests that violation data may be incomplete or unevenly recorded across vehicle types, an important consideration for statistical inference.
This regulatory transition creates a methodological challenge - comparisons between AVs and human drivers must account for potential enforcement bias, differences in citation practices, and uneven exposure across geographic regions.

**RESEARCH GAP:**

Finally, existing data infrastructures shape what types of comparative analyses are currently feasible. The National Highway Traffic Safety Administration’s Standing General Order (SGO) requires standardized reporting of crashes involving automated driving systems, providing a structured national dataset for AV safety analysis but excluding non-crash violations.<sup><a href="#fn7" id="ref7">7</a></sup> At the state level, California DMV disengagement reports offer detailed information about AV testing and deployment conditions, which researchers often use to study system limitations and environmental challenges, though these reports do not directly capture violations.<sup><a href="#fn8" id="ref8">8</a></sup> Together, prior work reveals a gap in the literature: while crashes and disengagements have been studied extensively, comparative traffic violation rates across weather conditions and vehicle types remain largely unexplored, motivating the present project.

higher-frequency behavioral metric).

By focusing on violation rates rather than crashes alone, this study provides a complementary perspective on autonomous vehicle safety - one centered on lawful behavior and regulatory compliance in everyday driving contexts, especially under adverse weather conditions.

### Important Metrics

**Weather Condition** (`weather_1`) is the central independent variable, allowing comparison of collision patterns across clear, cloudy, rainy, foggy, and snowy conditions. Snow occurs infrequently in California, which limits statistical power for that category.

**Road Surface Condition** (`road_surface`) complements weather by capturing residual precipitation effects such as wet roads after rain has stopped. Road surface directly affects braking distance and vehicle control.

**Collision Severity** (`collision_severity`) indicates crash seriousness and enables comparison of whether adverse weather leads to more severe outcomes.

**Victim Counts** (`killed_victims`, `injured_victims`) quantify the human impact of collisions beyond simple crash frequency.

**Primary Collision Factor** (`pcf_violation_category`) records the main violation category attributed by the investigating officer and serves as a proxy for unsafe driving behavior, helping distinguish driver-caused from environment-caused collisions.

**Collision Type** (`type_of_collision`) reveals whether certain crash patterns (e.g., rear-end, head-on) become more common under adverse weather.

**Date and Time** (`collision_date`, `collision_time`) enable temporal, seasonal, and time-of-day analysis.

**Location** (`county_location`, `latitude`, `longitude`) controls for regional variation in weather patterns, road infrastructure, and traffic density.

### Data Wrangling Summary

The raw SWITRS data was distributed as a SQLite database with four tables (`collisions`, `case_ids`, `parties`, `victims`), each loaded with a 150,000-row limit. These were left-joined on `case_id` into a single denormalized table. Columns with 90% or more missing values were dropped, along with internal helper columns. Date columns were converted to datetime, and numeric types were downcast (float64 to float32, int64 to int32) for memory efficiency. Count-based columns (killed, injured, party count) had missing values filled with zero, since blanks semantically represent zero events. Finally, fully-null rows and rows with implausible dates (before 1960 or after today) were removed, yielding a final cleaned dataset of 339,855 rows and 100 columns saved to `data/02-processed/switrs_clean.parquet`.

## Discussion and Conclusion

### Summary of Findings

After going through 339,855 SWITRS collision records, a few things stood out pretty clearly. First, the vast majority of collisions (~74%) happen in clear weather, which makes sense for California, but it also means our fog and snow samples are pretty small (fog is about 1 to 2% of records, and snow only has around 1,087 collisions total; see Figure 1). That said, even with those smaller samples, we found some interesting patterns.

Looking at the weather by violation-category heatmap (Figure 2), the types of violations that cause collisions shift noticeably depending on weather. Speeding jumps up a lot under fog (~47% of fog collisions) and snow (~73%), while traffic-signal violations go down. This lines up with what you'd expect: when visibility drops, speed becomes a bigger factor. Severity-wise (Figure 3), fog actually has the highest per-collision fatality rate in the whole dataset, even though there aren't that many fog collisions. So fog crashes are rare, but when they do happen, they tend to be really bad. The annual trend (Figure 4) shows the expected COVID-19 drop in 2020, and collisions generally peak around 3 to 6 PM rush hour regardless of weather (Figure 5). Geographically, collisions cluster in urban counties like LA, San Bernardino, and Orange (Figure 5), which are also the places that get the least severe weather.

On the AV side, the CA DMV Autonomous Vehicle Collision Reports (646 records, 2019 to 2024) from our EDA checkpoint tell a pretty different story. Most AV collisions are minor (73% minor damage, only about 2% major), and in roughly 84% of cases the AV was the one getting hit rather than causing the collision. Rear-end crashes are the most common type. The weather breakdown is even more lopsided than SWITRS (~84% clear), and there are literally zero snow collisions and only 3 to 5 fog cases in the entire dataset. This is basically because AV testing is concentrated in San Francisco, where it barely ever snows or gets dense fog. Unfortunately, this means we can't really test the most interesting part of our hypothesis: whether AV performance actually gets worse under the toughest weather conditions.

### How This Connects to Our Hypothesis

We hypothesized that AVs would have less severe collision outcomes under clear and rainy conditions, but that this advantage might shrink or flip under fog. The first part seems to hold up: AV collisions are mostly minor no matter the weather, and the fact that AVs are usually the ones getting hit (rather than causing crashes) suggests a lot of these incidents come down to human drivers around them, not the AV system itself messing up. But we really can't say much about the second part since there's almost no fog or snow data for AVs.

What we found in SWITRS does match what the literature on sensor degradation would suggest. The fact that fog has the highest fatality rate and that speeding violations spike under fog and snow (refs 3, 4) makes sense: low visibility gives drivers less room to recover from mistakes, so crashes that might have been fender-benders in clear weather end up being fatal. This is relevant for AVs too. If human drivers tend to speed more in bad visibility, then an AV driving cautiously in the same conditions is probably more likely to get rear-ended by someone following too close. The "struck by others" numbers from the AV data kind of support this idea, but honestly the sample sizes are way too small to draw any real conclusions from.

### Limitations

There are a number of limitations worth being upfront about:

- **No exposure data.** We don't have miles driven or traffic volume for either dataset, so we can't tell whether fog's higher fatality rate is because fog is genuinely more dangerous or just because fewer people drive in fog (and the ones who do might be taking riskier trips).
- **No AV flag in SWITRS.** SWITRS doesn't identify which vehicles are autonomous, so it's only useful as a human-driver baseline. Our AV comparison comes entirely from the separate DMV dataset, which uses a different severity scale (vehicle damage levels vs. injury categories), so comparing severity across the two isn't perfectly apples-to-apples.
- **Row-limit sampling.** We capped each SWITRS source table at 150,000 rows, which means the data skews toward around 2020. It doesn't evenly represent the full 2001 to 2020 timeframe.
- **Geographic concentration.** Most collisions in both datasets happen in urban California (LA, SF, San Diego), where severe weather is uncommon. Our results probably wouldn't generalize to places with harsher climates.
- **Reporting inconsistencies.** SWITRS covers dozens of police jurisdictions that don't all use the same reporting standards (refs 7, 8). And the AV dataset is self-reported by the manufacturers themselves, which is worth keeping in mind.
- **COVID-19 in 2020.** The big drop in collisions that year compresses our temporal baseline and could throw off trend comparisons.
- **Small fog and snow samples.** Snow has about 1,087 cases in SWITRS and zero in the AV data. Fog is similarly tiny on the AV side (3 to 5 cases). These are just too small for any serious statistical testing, especially if we wanted to look at interactions between weather and vehicle type.

### Next Steps

If we were to keep working on this, two things would help the most. First, fully integrating the AV collision dataset into the analysis pipeline alongside SWITRS so we can do proper matched comparisons by weather condition instead of just looking at the two datasets side by side. Second, getting vehicle-miles-traveled (VMT) data, whether from Caltrans estimates for human drivers or from AV company mileage disclosures, would let us calculate actual collision *rates* instead of just raw counts. That would go a long way toward addressing the biggest limitation of this whole analysis.

### Ethics

One thing we want to flag is that policy decisions about AVs (like whether to restrict them in bad weather) really shouldn't be made based on data as limited as what we have here. Without acknowledging the gaps, someone could use results like these to either over-restrict AVs, which could hurt communities that rely on autonomous ride services, or give them a premature green light for all-weather driving. It's also worth noting that collision data in general can carry systemic biases: underreporting tends to be worse in lower-income areas, enforcement varies a lot by neighborhood, and right now AV testing is mostly happening in wealthier urban areas. Being transparent about what the data actually shows, and what it doesn't, is important so that no one reads too much into these findings.
