## Abstract

As autonomous vehicles (AVs) move from controlled tests to everyday California roads, understanding how weather affects their collision patterns — and how those patterns compare to human drivers — is a pressing safety question. Rain, fog, and snow degrade the cameras and LiDAR that AVs rely on, yet most prior work focuses on crash rates rather than the types and severity of collisions that actually occur under different weather conditions.

This study uses two California-specific datasets to investigate that gap. The first is the Statewide Integrated Traffic Records System (SWITRS), which provides 339,855 police-reported human-driver collision records spanning 2001–2020 and serves as the human-driver baseline. The second is the California DMV Autonomous Vehicle Collision Reports dataset (646 records, 2019–2024), which documents AV-involved incidents. We conduct exploratory data analysis — including weather-by-violation heatmaps, row-normalized severity comparisons, temporal trend analysis, and hourly/geographic breakdowns — to characterize how collision patterns shift under adverse weather for each vehicle type.

Several findings stand out. For human drivers, clear weather accounts for roughly 74% of collisions, yet fog produces the highest per-collision fatality rate despite its small sample, and the share of speeding-related violations rises sharply under fog (47%) and snow (73%). Collision counts dropped steeply in 2020 due to COVID-19, and collisions peak between 3–6 PM with heavy geographic concentration in urban counties. For AVs, collisions are overwhelmingly low-severity (73% minor damage), approximately 84% involve the AV being struck by another party, and the dataset contains almost no fog or snow cases — limiting direct comparison under the most challenging conditions.

These results partially support the hypothesis that AVs exhibit less severe collision outcomes under normal and moderately adverse conditions, but the comparison arm under fog and snow remains untested due to data scarcity. A critical caveat is that no exposure data (e.g., miles driven) is available for either dataset, so observed counts reflect both true risk and driving activity. Future work should integrate exposure-normalized rates and geographically diverse AV deployment data to draw firmer conclusions.

## Research Question

How do weather conditions affect the characteristics and severity of collisions involving autonomous vehicles compared to human-driven vehicles in California?

Specifically, we examine whether autonomous vehicles operating in California are involved in different types or severities of collisions than human-driven vehicles under varying weather conditions such as clear weather, rain, fog, and snow. The independent variable is weather condition (clear, rain, fog, snow), the dependent variables are observed collision outcomes (including collision type, damage severity, and whether the vehicle was struck by another party), and the primary comparison variable is vehicle type (autonomous vs. human-driven).

Using California-specific traffic collision datasets, this study applies statistical inference methods to compare collision patterns across weather conditions and vehicle types. Because exposure measures such as miles driven are not available, the analysis focuses on differences in observed collision characteristics rather than true collision rates. By focusing on California, where autonomous vehicle testing and deployment are heavily regulated and documented, we aim to evaluate whether autonomous vehicles exhibit different safety-related collision patterns than human drivers under adverse weather conditions or whether their performance changes under certain environmental factors.



## Background and Prior Work

Autonomous vehicles (AVs) are increasingly deployed on public roads, raising important questions about their safety and compliance with traffic laws relative to human-driven vehicles. While much public discussion has focused on whether AVs reduce crash risk, less attention has been paid to how AVs behave with respect to traffic violations, such as speeding, red-light running, or improper lane changes. Weather conditions present a particularly relevant context for this comparison. Rain, fog, and snow degrade visibility and road friction, increasing driving difficulty for humans and simultaneously stressing the perception and control systems used by autonomous vehicles. Understanding how traffic violation rates vary by weather and vehicle type is therefore critical for evaluating whether autonomous vehicles maintain safer or more lawful behavior under adverse conditions.

**CRASH FOCUSED RESEARCH:**

Most existing research comparing autonomous and human-driven vehicles focuses on crash outcomes rather than traffic violations. A recent large-scale matched case–control study published in Nature Communications compared crash involvement between autonomous and human-driven vehicles while controlling for exposure factors such as location and time of day. This study found that AVs exhibited lower crash rates in certain contexts but emphasized that conclusions depend strongly on how exposure and confounding variables are handled.<sup><a href="#fn1" id="ref1">1</a></sup>

***Limitation:*** However, crash-based analyses face several limitations: crashes are relatively rare events, often underreported or inconsistently categorized, and may not capture everyday rule compliance or near-miss behavior. As noted by the National Highway Traffic Safety Administration (NHTSA), crash datasets are structured around reportable incidents and do not systematically capture non-crash traffic violations or routine enforcement actions.<sup><a href="#fn7" id="ref7">7</a></sup>

Similarly, Kusano et al. analyzed Waymo’s rider-only crash data and compared it to human-driver benchmarks, concluding that AV crash rates were comparable to or lower than human baselines under many conditions, though the authors noted limitations related to operating domains and reporting standards.<sup><a href="#fn2" id="ref2">2</a></sup> These studies establish methodological precedents, such as rate normalization and contextual matching, that are directly relevant for analyzing traffic violation rates, even though violations themselves were not the primary outcome. Because violations occur far more frequently than crashes, analyzing traffic citations may provide a higher-resolution measure of everyday driving behavior and regulatory compliance, particularly in varied environmental conditions.

**WEATHER AND SENSOR RESEARCH:**

A parallel body of work examines how adverse weather conditions affect autonomous vehicle perception and performance, providing insight into mechanisms that may influence traffic violations. Empirical studies of on-road autonomous systems have demonstrated that LiDAR and camera-based detection performance degrades measurably under rain, fog, and snow, leading to reduced object detection range and increased uncertainty.<sup><a href="#fn3" id="ref3">3</a></sup> Comprehensive review articles further document how adverse weather challenges sensor fusion, lane detection, and object tracking, and describe mitigation strategies such as radar integration and weather-robust perception models.<sup><a href="#fn4" id="ref4">4</a></sup> These findings suggest that even if AVs are conservative by design, degraded perception in poor weather may lead to behaviors that increase the likelihood of certain violations (e.g., improper lane positioning or failure to proceed through intersections efficiently).

***Limitation:*** However, these studies are largely conducted in controlled experimental settings or simulation environments. They typically measure detection range, object classification accuracy, or system disengagement frequency rather than legally defined traffic violations. As the California Department of Motor Vehicles (DMV) notes in its autonomous vehicle disengagement reporting program, disengagements reflect system limitations but do not necessarily indicate illegal driving behavior.<sup><a href="#fn8" id="ref8">8</a></sup> Thus, while perception degradation is documented, its real-world impact on enforceable traffic violations remains understudied.

**ENFORCEMENT AND REPORTING:**

Evidence that autonomous vehicles accumulate real-world traffic citations further motivates direct study of violations rather than crashes alone. In 2024, Waymo vehicles operating in San Francisco received hundreds of traffic tickets, primarily related to parking and obstruction, demonstrating that AVs are subject to, and do incur, formal violations under current enforcement regimes.<sup><a href="#fn5" id="ref5">5</a></sup>

***Limitation:*** While many of these citations are not moving violations, they highlight that AV behavior can conflict with traffic regulations in practice. Reporting has also noted ambiguity in how law enforcement handles moving violations committed by driverless vehicles, with California introducing new legislation (effective in 2026) to clarify enforcement procedures for autonomous systems.<sup><a href="#fn6" id="ref6">6</a></sup> This regulatory context suggests that violation data may be incomplete or unevenly recorded across vehicle types, an important consideration for statistical inference.
This regulatory transition creates a methodological challenge - comparisons between AVs and human drivers must account for potential enforcement bias, differences in citation practices, and uneven exposure across geographic regions.

**RESEARCH GAP:**

Finally, existing data infrastructures shape what types of comparative analyses are currently feasible. The National Highway Traffic Safety Administration’s Standing General Order (SGO) requires standardized reporting of crashes involving automated driving systems, providing a structured national dataset for AV safety analysis but excluding non-crash violations.<sup><a href="#fn7" id="ref7">7</a></sup> At the state level, California DMV disengagement reports offer detailed information about AV testing and deployment conditions, which researchers often use to study system limitations and environmental challenges, though these reports do not directly capture violations.<sup><a href="#fn8" id="ref8">8</a></sup> Together, prior work reveals a gap in the literature: while crashes and disengagements have been studied extensively, comparative traffic violation rates across weather conditions and vehicle types remain largely unexplored, motivating the present project.

**This project fills a distinct gap by:**

*   Shifting the outcome variable from crashes to traffic violations (a
higher-frequency behavioral metric).
*   Incorporating weather as a central explanatory variable rather than a background control.
* Using California-specific enforcement data to leverage the state’s unique concentration of AV deployment.

By focusing on violation rates rather than crashes alone, this study provides a complementary perspective on autonomous vehicle safety - one centered on lawful behavior and regulatory compliance in everyday driving contexts, especially under adverse weather conditions.

## Hypothesis


We hypothesize that autonomous vehicles will be associated with less severe collision outcomes and more predictable collision patterns than human-driven vehicles under clear and rainy conditions, but that this difference will narrow or potentially reverse under more challenging conditions such as fog or other low-visibility environments. This expectation is based on the idea that autonomous systems excel at rule-based driving and sensor fusion in normal or moderately adverse conditions, enabling consistent braking, lane maintenance, and hazard detection. However, their performance may degrade in difficult environmental conditions due to sensor limitations, such as reduced visibility affecting cameras and lidar systems. In contrast, human drivers may adapt more flexibly to unexpected environmental changes despite generally less consistent driving behavior. Because exposure measures such as miles driven are unavailable, this hypothesis concerns differences in observed collision characteristics and severity rather than true collision rates.

The raw SQLite database contains four relational tables (`collisions`, `case_ids`, `parties`, `victims`) that were merged on `case_id` to produce one row per collision-party-victim combination. SWITRS serves as the **human-driver baseline** against which AV collision patterns are compared.

### Important Metrics

**Weather Condition** (`weather_1`) is the central independent variable, allowing comparison of collision patterns across clear, cloudy, rainy, foggy, and snowy conditions. Snow occurs infrequently in California, which limits statistical power for that category.

**Road Surface Condition** (`road_surface`) complements weather by capturing residual precipitation effects such as wet roads after rain has stopped. Road surface directly affects braking distance and vehicle control.

**Collision Severity** (`collision_severity`) indicates crash seriousness and enables comparison of whether adverse weather leads to more severe outcomes.

**Victim Counts** (`killed_victims`, `injured_victims`) quantify the human impact of collisions beyond simple crash frequency.

**Primary Collision Factor** (`pcf_violation_category`) records the main violation category attributed by the investigating officer and serves as a proxy for unsafe driving behavior, helping distinguish driver-caused from environment-caused collisions.

**Collision Type** (`type_of_collision`) reveals whether certain crash patterns (e.g., rear-end, head-on) become more common under adverse weather.

**Date and Time** (`collision_date`, `collision_time`) enable temporal, seasonal, and time-of-day analysis.

**Location** (`county_location`, `latitude`, `longitude`) controls for regional variation in weather patterns, road infrastructure, and traffic density.

### Data Wrangling Summary

The raw SWITRS data was distributed as a SQLite database with four tables (`collisions`, `case_ids`, `parties`, `victims`), each loaded with a 150,000-row limit. These were left-joined on `case_id` into a single denormalized table. Columns with 90% or more missing values were dropped, along with internal helper columns. Date columns were converted to datetime, and numeric types were downcast (float64 to float32, int64 to int32) for memory efficiency. Count-based columns (killed, injured, party count) had missing values filled with zero, since blanks semantically represent zero events. Finally, fully-null rows and rows with implausible dates (before 1960 or after today) were removed, yielding a final cleaned dataset of 339,855 rows and 100 columns saved to `data/02-processed/switrs_clean.parquet`.

### Shortcomings

- **Collisions only**: excludes non-crash traffic violations, potentially underestimating risky behavior.
- **No AV indicator**: SWITRS does not identify autonomous vehicles; used only as a human-driver baseline.
- **No exposure data**: no miles driven or traffic volume; collision counts reflect both risk and driving activity.
- **Row-limit sampling**: 150k-row cap per table means this is a subset, primarily covering ~2020.
- **Geographic concentration**: severe weather is rare in California's urban areas where most collisions occur.
- **Reporting inconsistencies**: multiple jurisdictions with potentially different reporting standards.
- **Missing data**: core variables (`weather_1`, `collision_severity`) are >99% complete, but auxiliary fields (coordinates, collision time) have significant gaps.

### Dataset #2 California DMV Autonomous Vehicle Collision Reports (2019–2024)

**Dataset Name:** CA_AV_Collision_2019-2024.csv  
**Link:** https://github.com/xuyimingxym/CA-AV-Collision-Data  
**Number of observations:** 646  
**Number of variables:** 196 raw → 13 after cleaning

#### Important Metrics

**Weather** is recorded as a categorical variable with seven possible values: Clear, Cloudy, Raining, Snowing, Fog, Other, and Wind. This is the main independent variable in the analysis. The weather condition reflects what was reported at the time of the collision by the AV manufacturer on the DMV form. The distribution is highly imbalanced. About 84% of collisions occurred in clear weather, roughly 11% in cloudy conditions, and about 5% in rain. Fog and "Other" each account for fewer than 1% of cases. There are no recorded collisions in snow or wind conditions. This imbalance is important because most AV testing in California takes place in San Francisco where snow essentially doesn't occur and it is relatively rare for dense fog to occur during testing periods. Thus, the dataset contains very few observations in poor weather categories. Because there are only 5 fog cases and no snow cases at all, we really can't say much about how AVs perform in those conditions. Any conclusions drawn from such tiny (or nonexistent) samples would be weak and should be treated very cautiously.

**Autonomous Mode vs. Conventional Mode** is recorded using two binary columns (Yes or blank) to indicate whether the vehicle's self-driving system was in control at the time of the collision or whether a human safety driver was operating the vehicle. This is the central variable for comparing system-driven behavior to human-driven behavior within the same AV fleet. In this dataset, 358 collisions (55%) occurred while the vehicle was in autonomous mode and 284 (44%) occurred in conventional mode, with 4 records missing this information. The distribution is fairly balanced, which is helpful because it gives us a meaningful number of observations in both conditions. That said, it is important to interpret "conventional mode" correctly. In this context, it may refer to a trained safety driver employed by an AV company who is operating a vehicle equipped with autonomous technology. These drivers are typically more experienced, more attentive, and specifically trained for this role. As a result, comparisons between autonomous and conventional mode in this dataset likely underestimate the difference between AV performance and the broader driving population.

**Vehicle Damage** is recorded as a categorical variable with five levels: None, Minor, Moderate, Major, and Unknown. The damage level is reported by the AV company itself, following the DMV damage scale, rather than by an independent appraiser. "Minor" typically refers to cosmetic damage such as scratches or small dents with no structural impact. "Moderate" indicates that some structural components were affected, but the vehicle is still drivable. "Major" reflects significant structural damage and may mean the vehicle is no longer drivable. In this dataset, 73% of collisions resulted in minor damage, 15% in moderate damage, 9% in no damage at all, and only 2.3% in major damage. This distribution suggests that most AV-related collisions are relatively low-speed and low-severity incidents. For the purposes of our research question, damage severity acts as a practical measure of collision intensity. It allows us to examine whether certain conditions, such as adverse weather, are associated with more severe outcomes.

**Type of Collision** is recorded as a categorical variable with eight possible categories: Head-On, Sideswipe, Rear End, Broadside, Hit Object, Overturned, Vehicle/Pedestrian, and Other. This variable captures the physical layout or geometry of the crash. Rear-end collisions are the most common type. They account for 18% of cases from the AV's perspective and 32% from the other vehicle's perspective, which suggests that AVs are often being rear-ended rather than causing the rear-end impact themselves. This matters for the weather analysis because rear-end crashes are closely tied to following distance and braking behavior. In conditions like rain or on slippery roads, stopping distances increase and reaction time becomes more critical.

**Struck by Others** is a binary variable coded as 1 for yes and 0 for no. It indicates whether the AV was hit by another vehicle, rather than the AV striking something else. In this dataset, 84% of collisions involve the AV being the vehicle that was struck. For our research question, this variable helps separate collisions that may be linked to the AV's own driving behavior, which could potentially worsen in adverse weather, from collisions that are primarily the result of human driver behavior under the same conditions.

**Road Surface** is recorded as a categorical variable with four levels: Dry, Wet, Snowy/Icy, and Slippery. It describes the physical condition of the roadway at the time of the collision. Although it is related to the weather variable, it captures something slightly different. For example, the road may still be wet after rainfall has stopped, or it may remain dry during cloudy conditions. This variable complements the weather data by giving us a more precise way to assess how environmental factors influence collision patterns.

**Date of Accident and Time of Accident** are recorded separately as a date string (for example, "6/25/2021") and a time string (for example, "2:52"), along with separate AM/PM indicator columns. The dataset covers incidents from January 2019 through December 2024. Time is relevant because certain weather conditions like fog are more likely at specific times of day, and lighting conditions interact with weather to affect driving difficulty.

**Latitude and Longitude** are recorded in decimal degrees using the WGS84 coordinate system. Latitude ranges from about 32.66 (San Diego) up to 37.84 (northern San Francisco), while longitude ranges from roughly −122.51 to −116.27. The data are heavily concentrated around San Francisco (lat ≈ 37.77, lon ≈ −122.42), reflecting where most AV testing occurs. Location matters because weather patterns vary across regions, and we may need to account for geographic concentration to avoid attributing location-based effects to weather alone.

#### Concerns and Limitations

**Self-reported by AV companies.** Every record comes directly from the autonomous vehicle companies themselves. The company writes the accident summary, decides how severe the damage was, and reports whether the car was in autonomous or conventional mode. There is no outside investigator double-checking these details. AV manufacturers have obvious financial and reputational reasons to frame incidents favorably. So while 84% of collisions are reported as the AV being "struck by others," we should keep in mind that this conclusion is based on the company's own account.

**Extreme geographic concentration.** About 75% of the 646 records come from San Francisco, and nearly all remaining cases are from nearby Bay Area cities. San Francisco has a mild Mediterranean climate — it essentially never snows, dense street-level fog is uncommon, and rainfall is moderate. As a result, the dataset contains almost no observations for the weather conditions most relevant to our research question: snow has 0 records, fog has 3, and rain only 31. Any conclusions about AV performance in adverse weather will be based on very limited evidence, and results will not generalize to regions with harsher climates.

**No traffic violation data.** Our research question focuses on violation rates, but this dataset only records collisions, not violations. A collision is just one possible outcome of unsafe driving, and many violations never lead to a crash. We would need to treat collisions as an imperfect proxy for unsafe behavior or find an additional violation-specific dataset.

##### 2.2 — Damage Severity by Weather Condition

This visualization compares the distribution of damage severity across weather conditions for AV collisions. Each bar shows the percentage of collisions in that weather condition resulting in each damage level (rows sum to 100%).

However, direct comparison is limited because the SWITRS dataset uses injury-based severity categories while the AV dataset uses vehicle-damage categories. The small sample sizes for rain (n≈20) and fog (n≈3) also mean observed differences in those weather conditions may reflect sampling variability rather than true patterns.


```python
damage_order = ['None', 'Minor', 'Moderate', 'Major', 'Unknown']
ct_dmg = pd.crosstab(av['Weather'], av['Damage'], normalize='index').reindex(
    columns=[d for d in damage_order if d in av['Damage'].values]
) * 100

weather_order = av['Weather'].value_counts().index.tolist()
ct_dmg = ct_dmg.reindex(weather_order)

fig, ax = plt.subplots(figsize=(12, 5))
ct_dmg.plot.barh(stacked=True, ax=ax,
                  color=['#2ecc71', '#3498db', '#f39c12', '#e74c3c', '#95a5a6'][:ct_dmg.shape[1]])
ax.set_xlabel("Percentage of Collisions (%)")
ax.set_title("Figure 7 — Damage Severity by Weather Condition (row %)\n(AV Collisions, 2019–2024)",
             fontsize=12, fontweight='bold')
ax.legend(title="Damage Level", bbox_to_anchor=(1.02, 1), loc='upper left')

counts = av['Weather'].value_counts()
for i, weather in enumerate(ct_dmg.index):
    ax.text(101, i, f'n={counts[weather]}', va='center', fontsize=9, style='italic')

plt.tight_layout()
plt.savefig("results/fig7_av_damage_by_weather.png", dpi=150, bbox_inches="tight")
plt.show()

print("=== Damage severity by weather (row %) ===")
print(ct_dmg.round(1).to_string())
```

Rear-end collisions are the most common type across all weather conditions, which is consistent with the finding that ~91% of AV collisions involve the AV being struck by another party — suggesting that many AV collisions result from human drivers failing to anticipate AV braking behavior.

In Section 1, the human-driver heatmap showed that **speeding** was the dominant violation category across all weather conditions, with its share increasing in snow (73%) and fog (47%). Here, the AV collision type distribution is expected to be less varied because AVs are typically the struck party rather than the at-fault party.


```python
av_type = av.dropna(subset=['Collision Type 1'])
print(f"Rows with Collision Type 1 data: {len(av_type)} / {len(av)} ({len(av_type)/len(av)*100:.1f}%)")

ct_type = pd.crosstab(av_type['Weather'], av_type['Collision Type 1'], normalize='index') * 100
weather_order = av_type['Weather'].value_counts().index.tolist()
type_order = av_type['Collision Type 1'].value_counts().index.tolist()
ct_type = ct_type.reindex(index=weather_order, columns=type_order)

fig, ax = plt.subplots(figsize=(12, 5))
sns.heatmap(ct_type, annot=True, fmt='.1f', cmap='YlOrRd', ax=ax, linewidths=0.5,
            cbar_kws={'label': 'Row %'})
ax.set_title("Figure 8 — Weather × Collision Type (row %) — AV Collisions\n(2019–2024, non-missing only)",
             fontsize=12, fontweight='bold')
ax.set_ylabel("Weather Condition")
ax.set_xlabel("Collision Type")
plt.tight_layout()
plt.savefig("results/fig8_av_weather_collision_type.png", dpi=150, bbox_inches="tight")
plt.show()

print("\nSample sizes per weather (with collision type data):")
print(av_type['Weather'].value_counts().to_string())

```

This chart examines whether the rate at which AVs are **struck by other drivers** (versus the AV initiating contact) varies across weather conditions. This is a unique variable in the AV dataset that has no direct parallel in SWITRS. It provides insight into the direction of fault — if AVs are overwhelmingly struck by others, then most AV collisions may reflect human-driver error rather than AV system failures, regardless of weather.

### Combined Conclusions: Comparing Human-Driver and Autonomous Vehicle Collision Patterns

After examining both datasets in parallel, several key findings emerge:

**1. Weather Distribution Imbalance Limits Comparison**
Both datasets are dominated by clear-weather collisions, but the AV dataset is more extreme (~84% clear vs. ~74% for SWITRS). The AV dataset contains zero snow collisions and only 5 fog cases, making it impossible to test the hypothesis that AV performance degrades under the most challenging conditions. This reflects the geographic concentration of AV testing in San Francisco's mild climate.

**2. Collision Severity Differs Between Datasets**
Human-driver collisions (SWITRS) include fatal, severe injury, and property-damage-only outcomes across all weather conditions. Section 1's Figure 3 shows that fog and other adverse weather are associated with a slightly higher share of fatal/severe injury collisions for human drivers. AV collisions are overwhelmingly minor damage (~73%), with only ~2% major damage (Section 2, Figure 7). While this suggests AV collisions are less severe, the comparison is limited because SWITRS uses injury-based severity and the AV dataset uses vehicle-damage categories.

**3. Violation Types vs. Collision Types Tell Different Stories**
Section 1's heatmap (Figure 2) revealed that **speeding** dominates human-driver violation categories across all weather, with its share increasing dramatically under snow (73%) and fog (47%). For AVs, rear-end collisions are the dominant type, and ~84% of AV collisions involve the AV being struck by another party (Section 2, Figure 9). This suggests fundamentally different collision dynamics: human-driver collisions are often caused by human behavioral failures (speeding, improper turning), while AV collisions are more often caused by *other* human drivers failing to anticipate AV behavior.

**4. The "Struck by Others" Pattern**
The finding that AVs are predominantly the struck party across all weather conditions suggests that weather-related AV collision risk may be driven more by how human drivers around AVs behave in bad weather than by AV system failures themselves. This is an important nuance for interpreting weather effects on AV safety.

**5. Temporal Context**
Dataset 2 shows COVID-19 related dips in 2020 (Section 2, Figure 10). The AV dataset shows rapid growth in collision reports from 2021–2024 as commercial deployment scaled, meaning the data is not uniformly distributed over time.

**Regarding our hypothesis**: We hypothesized that AVs would show less severe collision outcomes under clear and rainy conditions but that this advantage might narrow under fog or snow. The data partially supports the first part — AV collisions are generally less severe across all observed weather conditions. However, we **cannot test** the second part because there are essentially no AV collision records under fog or snow. Additionally, because ~84% of AV collisions are caused by other drivers striking the AV, the severity comparison may reflect more about AV operating speed and environment than about AV system capability under adverse weather. Future analysis with geographically diverse AV deployment data and exposure-normalized rates would be needed to properly evaluate this hypothesis.

## Discussion and Conclusion

### Summary of Findings

After going through 339,855 SWITRS collision records, a few things stood out pretty clearly. First, the vast majority of collisions (~74%) happen in clear weather, which makes sense for California, but it also means our fog and snow samples are pretty small (fog is about 1 to 2% of records, and snow only has around 1,087 collisions total; see Figure 1). That said, even with those smaller samples, we found some interesting patterns.

Looking at the weather by violation-category heatmap (Figure 2), the types of violations that cause collisions shift noticeably depending on weather. Speeding jumps up a lot under fog (~47% of fog collisions) and snow (~73%), while traffic-signal violations go down. This lines up with what you'd expect: when visibility drops, speed becomes a bigger factor. Severity-wise (Figure 3), fog actually has the highest per-collision fatality rate in the whole dataset, even though there aren't that many fog collisions. So fog crashes are rare, but when they do happen, they tend to be really bad. The annual trend (Figure 4) shows the expected COVID-19 drop in 2020, and collisions generally peak around 3 to 6 PM rush hour regardless of weather (Figure 5). Geographically, collisions cluster in urban counties like LA, San Bernardino, and Orange (Figure 5), which are also the places that get the least severe weather.

On the AV side, the CA DMV Autonomous Vehicle Collision Reports (646 records, 2019 to 2024) from our EDA checkpoint tell a pretty different story. Most AV collisions are minor (73% minor damage, only about 2% major), and in roughly 84% of cases the AV was the one getting hit rather than causing the collision. Rear-end crashes are the most common type. The weather breakdown is even more lopsided than SWITRS (~84% clear), and there are literally zero snow collisions and only 3 to 5 fog cases in the entire dataset. This is basically because AV testing is concentrated in San Francisco, where it barely ever snows or gets dense fog. Unfortunately, this means we can't really test the most interesting part of our hypothesis: whether AV performance actually gets worse under the toughest weather conditions.

### How This Connects to Our Hypothesis

We hypothesized that AVs would have less severe collision outcomes under clear and rainy conditions, but that this advantage might shrink or flip under fog. The first part seems to hold up: AV collisions are mostly minor no matter the weather, and the fact that AVs are usually the ones getting hit (rather than causing crashes) suggests a lot of these incidents come down to human drivers around them, not the AV system itself messing up. But we really can't say much about the second part since there's almost no fog or snow data for AVs.

What we found in SWITRS does match what the literature on sensor degradation would suggest. The fact that fog has the highest fatality rate and that speeding violations spike under fog and snow (refs 3, 4) makes sense: low visibility gives drivers less room to recover from mistakes, so crashes that might have been fender-benders in clear weather end up being fatal. This is relevant for AVs too. If human drivers tend to speed more in bad visibility, then an AV driving cautiously in the same conditions is probably more likely to get rear-ended by someone following too close. The "struck by others" numbers from the AV data kind of support this idea, but honestly the sample sizes are way too small to draw any real conclusions from.

### Limitations

There are a number of limitations worth being upfront about:

- **No exposure data.** We don't have miles driven or traffic volume for either dataset, so we can't tell whether fog's higher fatality rate is because fog is genuinely more dangerous or just because fewer people drive in fog (and the ones who do might be taking riskier trips).
- **No AV flag in SWITRS.** SWITRS doesn't identify which vehicles are autonomous, so it's only useful as a human-driver baseline. Our AV comparison comes entirely from the separate DMV dataset, which uses a different severity scale (vehicle damage levels vs. injury categories), so comparing severity across the two isn't perfectly apples-to-apples.
- **Row-limit sampling.** We capped each SWITRS source table at 150,000 rows, which means the data skews toward around 2020. It doesn't evenly represent the full 2001 to 2020 timeframe.
- **Geographic concentration.** Most collisions in both datasets happen in urban California (LA, SF, San Diego), where severe weather is uncommon. Our results probably wouldn't generalize to places with harsher climates.
- **Reporting inconsistencies.** SWITRS covers dozens of police jurisdictions that don't all use the same reporting standards (refs 7, 8). And the AV dataset is self-reported by the manufacturers themselves, which is worth keeping in mind.
- **COVID-19 in 2020.** The big drop in collisions that year compresses our temporal baseline and could throw off trend comparisons.
- **Small fog and snow samples.** Snow has about 1,087 cases in SWITRS and zero in the AV data. Fog is similarly tiny on the AV side (3 to 5 cases). These are just too small for any serious statistical testing, especially if we wanted to look at interactions between weather and vehicle type.

### Next Steps

If we were to keep working on this, two things would help the most. First, fully integrating the AV collision dataset into the analysis pipeline alongside SWITRS so we can do proper matched comparisons by weather condition instead of just looking at the two datasets side by side. Second, getting vehicle-miles-traveled (VMT) data, whether from Caltrans estimates for human drivers or from AV company mileage disclosures, would let us calculate actual collision *rates* instead of just raw counts. That would go a long way toward addressing the biggest limitation of this whole analysis.

### Ethics

One thing we want to flag is that policy decisions about AVs (like whether to restrict them in bad weather) really shouldn't be made based on data as limited as what we have here. Without acknowledging the gaps, someone could use results like these to either over-restrict AVs, which could hurt communities that rely on autonomous ride services, or give them a premature green light for all-weather driving. It's also worth noting that collision data in general can carry systemic biases: underreporting tends to be worse in lower-income areas, enforcement varies a lot by neighborhood, and right now AV testing is mostly happening in wealthier urban areas. Being transparent about what the data actually shows, and what it doesn't, is important so that no one reads too much into these findings.
